###Visualisation 1: Buyers of New Homes

**Source:** Housing In London 2025 Dataset - London Data Store

Sheet 2.6 titles removed, converted to CSV.

https://data.london.gov.uk/dataset/housing-in-london-the-evidence-base-for-the-mayors-housing-strat-24rpx

```python
#import libraries
import matplotlib.pyplot as plt
import pandas as pd

#setting data frame
df = pd.read_csv("Data & Resources/london_new_home_buyers-1.csv")


total_row = df[df['buyer_type'] == 'Total'].iloc[0]

#extract timeframes and values
quarters = df.columns[1:]
values = [total_row[q] for q in quarters]

#plot graph
plt.figure(figsize=(12, 6))
plt.bar(quarters, values, color='#000000', alpha=0.7, edgecolor='black')
plt.xlabel('Quarter')
plt.ylabel('Number of Buyers')
plt.title('Total New Home Buyers in London')
plt.xticks(rotation=90, ha='right')

plt.show()
```

###Visualisation 2: Relative Poverty in London

**Source:** London Datastore

xls -> csv

https://data.london.gov.uk/dataset/hbai-poverty-in-london-e79lj

```python
#import libraries
import matplotlib.pyplot as plt
import pandas as pd

#set dataframe
df = pd.read_csv("Data & Resources/relative_poverty-1.csv", skiprows=4)
#Claude used to help learn how to reformat to be readable in line below
df.rename(columns={df.columns[0]: 'Year'}, inplace=True)


#defining the axis and focusing on stats for all people with before and after ahter housing costs
year = df['Year'].values
all_people_bhc = pd.to_numeric(df.iloc[:, 1], errors='coerce')
all_people_ahc = pd.to_numeric(df.iloc[:, 4], errors='coerce')

#plotting the graph
plt.figure(figsize=(14, 6))
plt.plot(year, all_people_bhc, color = 'g', marker='s', label='Before Housing Costs')
plt.plot(year, all_people_ahc, color = 'b', marker='s', label='After Housing Costs')
plt.xlabel('Period', fontsize=12)
plt.ylabel('Relative Poverty Rate (%)', fontsize=12)
plt.title('Relative Poverty: All People in UK', fontsize=14)
plt.xticks(rotation=90, ha='right')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
