## Task 2: Create a scatter plot

a) Create a scatter plot using the "cars" dataset to visualise the relationship between "Horsepower" and "Miles Per Gallon. Set the colour of the points to a specific RGB triplet (e.g.,'rgb(255, 0, 0)' for red).

```python
#setup
cars = data.cars()

#find the table details
cars.head(10)
```

```python
#scatter plot of
cars_scatter = alt.Chart(cars).mark_point(filled=True).encode(
    x='Horsepower',
    y='Miles_per_Gallon',
    color = 'Origin',
    #tooltip to show the car's name when point hovered
    tooltip = 'Name'
).properties(
    title = 'Scatter Plot of Horsepower vs Miles Per Gallon'
)

cars_scatter
```

## Task 3: Create a line plot

a) Create a line plot to visualise the trend of maximum temperature over time in the "seattle_weather" dataset. Set the background to "grey" and the lines to "black".

```python
#setup
seattle_weather = data.seattle_weather()


#find the table details
seattle_weather.head(10)
```

```python
#line graping the above weather
seattle_weather_line = alt.Chart(seattle_weather).mark_line().encode(
    x = alt.X('date', axis=alt.Axis(gridColor='black')),
    y = alt.Y('temp_max', axis=alt.Axis(gridColor='black')),
    color = 'weather',
    tooltip = 'date'
).properties(
    background = 'lightgrey'
)

seattle_weather_line
```

## Task 4: Create a bar chart

Create a bar chart to visualise the distribution of population across different age groups in the "population" dataset.

```python
#setup
population = data.population()

#find the table details
population.head(10)
```

```python
#plottig a bar chart of population
population_bar = alt.Chart(population).mark_bar().encode(
    alt.X('age:O'),
    alt.Y('people:Q'),
    color = 'year',
    tooltip = 'year'
)

population_bar
```

Create a stacked bar chart to visualise the contribution of different energy sources (e.g., fossil, renewables) to the total net electricity generation in Iowa (dataset: "iowa_electricity") over time.

## Task 6: Create a streamgraph

Create a streamgraph to visualise the changing proportions of unemployment across the "Education and Health" and "Information" sectors over time using the "unemployment_across_industries" dataset.

```python
#setup
unemployment_across_industries = data.unemployment_across_industries()

#find the table details
unemployment_across_industries.head(10)
```

```python
#streamgraph of unemployment for specific sectors
unemployment_stream = alt.Chart(unemployment_across_industries).transform_filter(
    alt.FieldOneOfPredicate(field='series', oneOf=['Education and Health', 'Information'])
).mark_area().encode(
    alt.X('date:T'),
    alt.Y('rate:Q', stack='center'),
    alt.Color('series:N'),
    tooltip = ['rate', 'date']
)

unemployment_stream
```

## Task 7: Create an indexed line chart

Create an indexed line chart to visualise the relative price movements of different stocks over time using the "stocks" dataset.

```python
#setup
stocks = data.stocks()

#find the table details
stocks.head(10)
```

## Task 7: Create a heatmap

Create a heatmap to visualise the relationship between flight distance and average arrival delay in the "flights_5k" dataset. Try out different colour schemes.

```python
#setup
flights = data.flights_5k()

#find the table details
flights.head(10)
```

```python
#heatmapping flight distance and arrival delay
delay_distance_heatmap = alt.Chart(flights).mark_rect().encode(
    alt.Y('distance:Q').bin(maxbins=30),
    alt.X('delay:Q').bin(maxbins=50),
    alt.Color('count()').scale(scheme='greenblue'),
    tooltip = ['count()']
)

#putting a veritcal line to show where delays and early leaves seperate
zero_delay_line = alt.Chart(pd.DataFrame({'delay': [0]})).mark_rule(
    color='red',
    strokeWidth=2
    ).encode(
    x='delay'
)

alt.layer(delay_distance_heatmap, zero_delay_line)
```

## Task 8: Create a parallel coordinates plot

Create a parallel coordinates plot to visualise the relationships between sepal length, sepal width, petal length, and petal width for different species of Iris flowers in the "iris" dataset.

```python
#setup
iris = data.iris()

#find the table details
iris.head(10)
```

```python
#coordinates parallel of iris dataset
iris_parallel = alt.Chart(iris, width = 500).transform_window(
    index='count()'
).transform_fold(
    ['sepalLength', 'sepalWidth', 'petalLength', 'petalWidth']
).mark_line().encode(
    x = 'key:N',
    y = 'value:Q',
    color = 'species:N',
    detail = 'index:N',
    tooltip = 'value:Q',
    opacity=alt.value(0.5)
)

iris_parallel
```
